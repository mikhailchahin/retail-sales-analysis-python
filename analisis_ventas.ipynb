```python
#Extrae el numero de mes
df['Date'].dt.month
```

```python
#Cuenta la cantidad de ventas por año
df['Date'].dt.year.value_counts()
```

```python
ventas_por_mes = df.groupby(df["Date"].dt.month)["Total Amount"].sum() 
```

```python
print(ventas_por_mes.index)
```

```python
import calendar 
def convertir_mes(mes):
    nombre = calendar.month_name[mes]   
    return nombre
```

```python
ventas_por_mes.index = ventas_por_mes.index.map(convertir_mes)
ventas_por_mes.sort_values(ascending=False)
print(ventas_por_mes)
```

```python
plt.figure(figsize=(9, 7))

plt.bar(ventas_por_mes.index, ventas_por_mes.values, color='skyblue')

plt.title('Monthly sales', fontsize=16)
plt.xlabel('Month', fontsize=14)      
plt.ylabel('Total Sales', fontsize=14)
plt.xticks(rotation=45)

# Agregar los importes encima de las barras
for i,value in enumerate(ventas_por_mes.values):
    plt.text(i, value, f'{value:,.0f}', ha='center', va='bottom', fontsize=10)

plt.show()
```

```python
q1 = df["Total Amount"].quantile(0.25) 
q3= df["Total Amount"].quantile(0.75)

iqr = q3 - q1
print(f"IQR: {iqr}")
```

```python
multiplicador = 1.5 * iqr
print(f"Multiplicador: {multiplicador}")
```

```python
limite_inferior = q1 - multiplicador
limite_superior = q3 + multiplicador

print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)
```

```python
outliers = df[df["Total Amount"] > limite_superior]

print(outliers)
```

```python
#Venta mas alto valor
df["Total Amount"].max()
```

```python
cantidad_outliers = len(outliers)

print("Cantidad de outliers:", cantidad_outliers)
```

```python
# Establecemos el tamaño del gráfico (ancho: 8, alto: 5)
plt.figure(figsize=(8, 5))

# Creamos un Boxplot utilizando los importes de las ventas
plt.boxplot(df["Total Amount"])

# Agregamos el título del gráfico
plt.title("Sales Distribution - Boxplot")

# Agregamos el nombre al eje vertical
plt.ylabel("Total Amount")

# Mostramos el gráfico en pantalla
plt.show()
```

```python
# Agrupamos las ventas según la categoría del producto
# y sumamos el importe total de cada categoría
ventas_categoria = df.groupby("Product Category")["Total Amount"].sum()

# Ordenamos las categorías de mayor a menor facturación
ventas_categoria = ventas_categoria.sort_values(ascending=False)

# Mostramos el resultado
print(ventas_categoria)
```

```python
# Creamos una figura y establecemos su tamaño
plt.figure(figsize=(9, 5))

# Creamos un gráfico de barras con las categorías y sus ventas
plt.bar(ventas_categoria.index, ventas_categoria.values)

# Agregamos el título del gráfico
plt.title("Total Sales by Product Category")

# Etiquetamos el eje horizontal
plt.xlabel("Product Category")

# Etiquetamos el eje vertical
plt.ylabel("Total Sales")

# Mostramos el gráfico
plt.show()
```

```python
# Creamos una figura y establecemos su tamaño
plt.figure(figsize=(9, 5))

# Creamos el histograma utilizando los importes de las ventas
# bins=10 divide los datos en 10 intervalos
plt.hist(df["Total Amount"], bins=10)

# Agregamos el título
plt.title("Distribution of Sales Amounts")

# Etiquetamos el eje horizontal
plt.xlabel("Total Amount")

# Etiquetamos el eje vertical
plt.ylabel("Frequency")

# Mostramos el histograma
plt.show()
```

```python
# Calculamos la desviación estándar muestral con NumPy
# ddof=1 utiliza el mismo criterio que Pandas por defecto
desviacion = np.std(df["Total Amount"], ddof=1) # ddof=1 para obtener la desviación estándar muestral

# Mostramos el resultado
print(desviacion)
```
